Free Space Path Loss (FSPL) 

```python
import numpy as np
```

```python
def fspl(frequency: float, distance: float) -> float:
    fspl = 32.45 + 20 * np.log10(frequency) + 20 * np.log10(distance)
    return float(fspl)
```

```python
fspl(2, 100)
```

```python
def received_power_friis_equation(transmit_power: float, transmit_gain: float, transmit_loss: float, path_loss: float, receive_gain: float, receive_loss: float) -> float:
    """this function calculates the received power using the Friis transmission equation"""
    receiver_power = np.log10(transmit_power) + np.log10(transmit_gain) - np.log10(transmit_loss) - np.log10(path_loss) + np.log10(receive_gain) - np.log10(receive_loss)
    return float(receiver_power)
```

```python
received_power_friis_equation(100, 2, 1, 50, 2, 1)
```

```python
def link_margin(required_power: float) -> float:
    """this function calculates the link margin"""
    link_margin = received_power_friis_equation(100, 2, 1, 50, 2, 1) - np.log10(required_power)
    return link_margin
```

```python
def eirp(transmit_power: float, transmit_gain: float, transmit_loss: float) -> float:
    """this function calculates the effective isotropic radiated power (EIRP)"""
    eirp = np.log10(transmit_power) + np.log10(transmit_gain) - np.log10(transmit_loss)
    return float(eirp)
```

```python
def effective_aperture(receiver_gain: float, wavelength: float) -> float:
    """this function calculates the effective aperture of the receiver"""
    effective_aperture = (np.log10(receiver_gain) * (wavelength ** 2)) / (4 * np.pi)
    return float(effective_aperture)
```
